```python
import pandas as pd
import numpy as np

certs = pd.DataFrame(
    {
        "uprn": [1001, 1002, 1001, 1003, 1002, 1003],
        "inspection_date": [
            "2022-04-01",
            "2024-01-10",
            "2025-06-15",
            "2023-08-20",
            "2021-03-05",
            "2025-01-12",
        ],
        "score": [55, 68, 82, 44, 35, 73],
    },
    index=[50, 10, 90, 20, 5, 70]
)

certs["inspection_date"] = pd.to_datetime(
    certs["inspection_date"]
)

certs
```

```python
latest = (
    certs.sort_values(
        by=["uprn", "inspection_date"],
        ascending=[True, False]
    )
    .drop_duplicates(
        subset=["uprn"],
        keep="first"
    )
)

print(latest.to_string())
```

```python
assert latest["uprn"].is_unique
```

```python
max_scores = certs.groupby("uprn")["score"].max()
print(max_scores.to_string())
```

```python
conditions = [
    certs["score"] >= 70,
    certs["score"] >= 40,
]

choices = [
    "Good",
    "Average"
]


certs["band"] = np.select(
    condlist=conditions,
    choicelist=choices,
    default="Poor"
)
```

```python
certs["property_max_score"] = certs["uprn"].map(max_scores)

print(certs.to_string())
```
